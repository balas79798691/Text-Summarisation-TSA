# Smart Text Summarizer

An application that converts lengthy text into a shorter extractive summary.

## 1. Install and Import

```python
!pip -q install sumy
import nltk
import re
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lsa import LsaSummarizer

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
print("Smart Text Summarizer is ready.")
```

## 2. Preprocessing

```python
def normalize_text(text):
    text = re.sub(r"\s+", " ", text)
    return re.sub(r"\s+([,.!?])", r"\1", text).strip()

def sentence_count(text):
    return len(nltk.sent_tokenize(normalize_text(text)))
```

## 3. Summarization Engine

```python
def summarize_text(text, number_of_sentences=3):
    text = normalize_text(text)
    parser = PlaintextParser.from_string(text, Tokenizer("english"))
    summarizer = LsaSummarizer()
    summary = summarizer(parser.document, number_of_sentences)
    return [str(sentence) for sentence in summary]

sample = (
    "Artificial intelligence is changing many areas of modern life. "
    "It is used in healthcare to assist doctors with diagnosis and research. "
    "Businesses use artificial intelligence to analyze large amounts of data. "
    "Education systems can provide personalized learning experiences. "
    "Machine learning allows systems to learn patterns from data. "
    "Organizations must also consider privacy, security, fairness, and responsible use."
)

for sentence in summarize_text(sample, 3):
    print(sentence)
```

## 4. Application

```python
def text_summarizer():
    print("=" * 60)
    print("             SMART TEXT SUMMARIZER")
    print("=" * 60)
    print("Paste text line by line. Type END when finished.\n")

    lines = []
    while True:
        line = input()
        if line.strip().upper() == "END":
            break
        lines.append(line)

    text = " ".join(lines)

    if len(text.split()) < 30:
        print("Please provide at least 30 words.")
        return

    value = input("Summary sentences [3]: ")
    try:
        count = int(value) if value.strip() else 3
        count = max(1, min(count, sentence_count(text)))
    except ValueError:
        count = 3

    print("\nGENERATED SUMMARY")
    print("-" * 60)
    for sentence in summarize_text(text, count):
        print(sentence)

# Uncomment to run:
# text_summarizer()
```

## 5. Compression Analysis

```python
summary = summarize_text(sample, 3)
original_words = len(sample.split())
summary_words = len(" ".join(summary).split())
reduction = 100 * (1 - summary_words / original_words)

print("Original words :", original_words)
print("Summary words  :", summary_words)
print("Reduction      :", round(reduction, 2), "%")
```

## Conclusion

The application uses extractive LSA summarization to reduce lengthy passages while retaining important sentences.
